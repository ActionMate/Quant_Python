#### 4. Factorial

```python
n=int(input('Enter number you want factorial of : '))
c=1
for i in range(1,n+1):
    c*=i
print('factorial : ',c)
```

#### 7. Base Conversion of Decimal to Binary/Octal/Hexadecimal

```python
n = int(input("Enter a decimal number: "))

# Decimal to Binary
num = n
binary = ""
if num == 0:
    binary = "0"
else:
    while num > 0:
        r = num % 2
        binary = str(r) + binary
        num = num // 2

# Decimal to Octal
num = n
octal = ""
if num == 0:
    octal = "0"
else:
    while num > 0:
        r = num % 8
        octal = str(r) + octal
        num = num // 8

    # Decimal to Hexadecimal
num = n
hexadecimal = ""
digits = "0123456789ABCDEF"
if num == 0:
    hexadecimal = "0"
else:
    while num > 0:
        r = num % 16
        hexadecimal = digits[r] + hexadecimal
        num = num // 16

    print("Binary      :", binary)
    print("Octal       :", octal)
    print("Hexadecimal :", hexadecimal)
```
