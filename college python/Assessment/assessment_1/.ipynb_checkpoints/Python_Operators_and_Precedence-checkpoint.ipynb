## 6. Logical Operators
#### Definition: Used to combine or modify Boolean conditions.

##### 1. Logical AND ( and )

```python
has_ticket = True
has_id = True
has_id_2 = False 

# Prints True only if both conditions are met
can_enter = has_ticket and has_id
can_enter_2 = has_ticket and has_id_2

print(can_enter)
print(can_enter_2)
```

##### 2. Logical OR  ( or )

```python
has_ticket = True
has_id = True
has_id_2 = False 

# Prints True if one of the conditions are met
can_enter = has_ticket or has_id
can_enter_2 = has_ticket or has_id_2

print(can_enter)
print(can_enter_2)
```

#### 3. Logical NOT  ( not )

```python
is_raining = True

# Flips True to False
go_outside = not is_raining
print(go_outside)  
```

## 9. Control Flow Statements 
#### Definition: Used to control the order in which statements are executed in a program.

##### 1. if Statement 
###### Definition: Executes a block of code only when a specified condition is True.

```python
a = 20
if a >= 18:
    print("You win")
```

##### 2. if-else Statement  ( if followed by else )
###### Definition: Chooses between two blocks of code depending on whether a condition is True or False.

```python
a = 10
if a >= 18:
    print("You win")
else :
    print("you lose")
```

##### 3. if-elif-else statement ( to use multiple if's )
###### Definition: Tests multiple conditions sequentially and executes the block belonging to the first condition that is True.

```python
a = 16
if a >= 18:
    print("You win")
elif a >= 12:
    print("Draw")
else:
    print("You lose")
```

##### 4. for Loop Statement
###### Definition: Repeats a block of code for each item in an iterable.

```python
for n in range(1, 6):     #range : goes from 1 to (last term -1), eg. in this case : 1,2,3,4,5
    print(n)
```

##### 5. while Loop 
###### Definition: Repeatedly executes a block of code as long as a condition remains True.

```python
n = 1
while n <= 5:
    print(n)
    n += 1
```

##### 6. break 
###### Definition: Immediately terminates the current loop.

```python
for n in range(1, 10):
    if n == 5:
        break
    print(n)
```

##### 7. continue 
###### Definition: Skips the current iteration of a loop and moves to the next iteration.

```python
for n in range(1, 6):
    if n == 3:
        continue
    print(n)
```

##### 8. pass
###### Definition: Does nothing and is used as a placeholder where a statement is syntactically required.

```python
for n in range(1, 6):
    if n == 2:
        pass
    print(n)
```

## 11. Operator Precedence

#### Definition: Operator precedence determines the order in which Python evaluates different operators in an expression.

### Why is operator precedence important?

##### When an expression contains multiple operators, Python does not evaluate them simply from left to right. It follows a specific order of precedence.

```python
a = 10 + 5 * 2
print(a)
```

##### explanation : Here, multiplication is performed before addition:
#####      10 + 5 * 2 => 10 + 10 => 20

```python
a = (10 + 5) * 2
print(a)
```

##### Parentheses have higher precedence, so the expression inside the parentheses is evaluated first.
##### (10 + 5) * 2 => 15 * 2 => 30

### Basic Order of Operator Precedence

| Priority | Operators / Expressions |
|----------|-------------------------|
| 1 | Parentheses `( )` |
| 2 | Exponentiation `**` |
| 3 | Unary `+`, `-`, `~` |
| 4 | Multiplication, Division, Floor Division, Modulus `*`, `/`, `//`, `%` |
| 5 | Addition and Subtraction `+`, `-` |
| 6 | Comparison operators `==`, `!=`, `>`, `<`, `>=`, `<=` |
| 7 | Logical `not` |
| 8 | Logical `and` |
| 9 | Logical `or` |

### Examples of Operator Precedence

```python
a = 10 + 2 * 3
print(a)
```

```python
a = (10 + 2) * 3
print(a)
```

```python
a = 2 + 3 ** 2
print(a)
```

```python
a = 20 - 6 / 2 + 3
print(a)
```

```python
a = 10
b = 5

result = a > b and b > 2
print(result)
```

### Changing Precedence Using Parentheses

```python
a = 10 + 5 * 2
b = (10 + 5) * 2

print(a)
print(b)
```

```python
a = 2 + 3 * 4 ** 2
print(a)
```
